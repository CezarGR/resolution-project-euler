```python
def sum_natural_number_multiple_numbers_three_and_five(target_num):
    count = 1
    result = 0
    while count < target_num:
        if count % 3 == 0 or count % 5 == 0:
            result += count
            
        count += 1

    return result
```

```python
%%timeit
sum_natural_number_multiple_numbers_three_and_five(1000)
```

```python
def sum_natural_number_multiple_numbers_three_and_five(target_num):
    m_3 = (target_num - 1) // 3
    m_5 = (target_num - 1) // 5
    m_15 = (target_num - 1) // 15
    
    return int((((1 + m_3) * (m_3/2)) * 3) + ((1 + m_5) * (m_5/2) * 5) - ((1 + m_15) * (m_15/2) * 15))
```
