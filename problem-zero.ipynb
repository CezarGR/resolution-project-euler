```python
def sum_odd_squares_by_quantity(quantity):
    return sum([ n**2 for n in range(1, ((quantity//2)*2), 2)])
```

```python
sum_odd_squares_by_quantity(916_000)
```

```python
def sum_odd_squares_by_quantity_2(quantity):
    result = (quantity * (4*(quantity/2)**2 - 1))/3
    return int(result)
```

```python
sum_odd_squares_by_quantity_2(916_000)
```
