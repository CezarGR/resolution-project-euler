```python
def calculate_fibonacci_sequence_with_limit_number(limit_number):
    a = 0
    b = 1
    result = 0

    while True:
        a, b = b, a+b

        if b > limit_number:
            break

        if b%2 == 0:
            print(b)
            result += b
    
    return result
```
